```python
import pandas as pd

```

```python
df = pd.read_parquet("./yellow_tripdata_2026-01.parquet")
```

```python
df.head()
```

```python
print("NA: ", '\n', df.isna().sum())
```

```python
mask = (
    df["passenger_count"].isna() 
)
passenger_count = df.loc[~mask]["passenger_count"].sum()
mean = passenger_count / 2636831
mean
df.loc[mask, "passenger_count"] = mean
```

```python
median = df["RatecodeID"].median()
df.loc[df["RatecodeID"].isna(), "RatecodeID"] = median
df["RatecodeID"].isna().sum()
```

```python
df["congestion_surcharge"].median()
df.loc[df["congestion_surcharge"].isna(), "congestion_surcharge"] = 2.5
```

```python
df["Airport_fee"].median()
df.loc[df["Airport_fee"].isna(), "Airport_fee"] = 0.0
```
